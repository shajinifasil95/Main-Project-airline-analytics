**Dataset Column/features Description:**

* **MEMBER_NO**-Unique membership number for each airline customer.
  **FFP_DATE**-Date when the customer joined the Frequent Flyer Program (FFP).
  **FIRST_FLIGHT_DATE**-Date of the customer's first recorded flight.
  **GENDER**-Gender of the customer.
  **FFP_TIER**-Frequent Flyer Program membership level or loyalty tier.
  **WORK_CITY**-City where the customer works.
  **WORK_PROVINCE**-Province/state where the customer works.
  **WORK_COUNTRY**-Country where the customer works.
  **AGE**-Age of the customer.
  **LOAD_TIME**-Date/time representing the data observation or analysis period.
  **FLIGHT_COUNT**-Total number of flights taken by the customer.
  **BP_SUM**-Total accumulated points associated with flight bookings/benefits.
  **SUM_YR_1**-Customer's total flight-related expenditure/revenue during Year1.
  **SUM_YR_2**-Customer's total flight-related expenditure/revenue during Year2.
  **SEG_KM_SUM**-Total distance travelled by the customer, measured in kilometers.
  **LAST_FLIGHT_DATE**-Date of the customer's most recent flight.
  **LAST_TO_END**-Number of days since the customer's last flight.
  **AVG_INTERVAL**-Average number of days between the customer's flights.
  **MAX_INTERVAL**-Maximum number of days between two flights.
  **EXCHANGE_COUNT**-Number of times the customer exchanged/redeemed points.
  **avg_discount**-Average discount received by the customer.
  **Points_Sum**-Total points accumulated by the customer.
  **Point_NotFlight**-Number of points earned through non-flight activities.

```python
skewed_cols = [
    'BP_SUM',
    'SUM_YR_1',
    'SUM_YR_2',
    'SEG_KM_SUM',
    'POINTS_SUM',
    'AVG_INTERVAL',
    'MAX_INTERVAL',
    'EXCHANGE_COUNT',
    'POINTS_NOTFLIGHT'
]

for col in skewed_cols:
    df[col + '_log'] = np.log1p(df[col])
```

```python
print(df.columns.tolist())
```

```python
Q1 = df['BP_SUM_log'].quantile(0.25)
Q3 = df['BP_SUM_log'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

outliers = df[
    (df['BP_SUM_log'] < lower) |
    (df['BP_SUM_log'] > upper)
]

print("Number of outliers:", len(outliers))
```

```python
#check IQR capping with log transformed columns
#capping limits extreme value without deleting
```

```python
Q1 = df['SEG_KM_SUM_log'].quantile(0.25)
Q3 = df['SEG_KM_SUM_log'].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

df['SEG_KM_SUM_log'] = df['SEG_KM_SUM_log'].clip(
    lower=lower,
    upper=upper
)
```

```python
#check IQR capping for all log transformed columns
```

```python
for col in skewed_cols:
    log_col = col + '_log'

    Q1 = df[log_col].quantile(0.25)
    Q3 = df[log_col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    df[log_col] = df[log_col].clip(
        lower=lower,
        upper=upper
    )
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(15, 7))
sns.boxplot(data=df[[col + '_log' for col in skewed_cols]])

plt.xticks(rotation=45)
plt.title('Box Plot After Log Transformation and Outlier Capping')
plt.show()
```
